Ejercicio 2. Escribir una funcion que simule un volado  

```python
def lanzamiento():
  u = r.random()
  # print (f"El aleatorio es {u}")
  if 0 <= u < 0.5:
    return "Sol"
  else:
    return "Aguila "
print ("Resultado: ",lanzamiento())
```

Ejercicio 3. Escribir una funcion que simule un volado y comprobar la ley de los grandes numeros. Usa una grafica para comparar

```python
def lanzar_moneda():
    u = r.random()
    if 0 <= u < 0.5:
        return 1 # Sol
    else:
        return 0 # Águila
```

```python
num_lanzamientos = 1000
resultados_suma = 0
probabilidades = []
```

```python
for i in range(1, num_lanzamientos + 1):
    resultado = lanzamiento()
    if resultado == "Sol":
        resultados_suma += 1
    promedio = resultados_suma / i
    probabilidades.append(promedio)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(probabilidades, label='Probabilidad empirica')
plt.axhline(y=0.5, color='r', linestyle='--', label='Probabilidad teorica')
plt.title('Ley de los grandes numeros')
plt.xlabel('Lanzamientos')
plt.ylabel('Probabilidad')
plt.legend()
plt.grid(True)
plt.show()
```
